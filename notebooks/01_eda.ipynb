# Exploratory Data Analysis — Disaster Tweets

This notebook is purely for understanding the data. No training happens here.

```python
import sys
sys.path.insert(0, '..')  # so we can import from src/

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from src import evaluate

train = pd.read_csv('../data/raw/train.csv')
test  = pd.read_csv('../data/raw/test.csv')

print(f"Train: {train.shape}, Test: {test.shape}")
train.head()
```

## Class distribution

```python
ax = train['target'].value_counts().plot(kind='bar', color=['steelblue', 'salmon'], rot=0)
ax.set_xticklabels(['Not Disaster (0)', 'Disaster (1)'])
ax.set_title('Class Distribution')
ax.set_ylabel('Count')
plt.tight_layout()
plt.show()

print(train['target'].value_counts(normalize=True).round(3))
```
